## Objectif
Cette semaine, nous allons découvrir la bibliothèque Pandas et apprendre à manipuler des données sous forme de tableaux.

## 1. Importation de Pandas

Pandas est une bibliothèque Python qui permet de manipuler et d'analyser des données.
On l'importe avec le raccourci `pd`.

```python
import pandas as pd
```

## 2. Chargement du fichier CSV

La fonction `pd.read_csv()` permet de lire un fichier CSV et de stocker ses données dans un DataFrame.

```python
df=pd.read_csv("../messages.csv")
```

## 3. Exploration du DataFrame

- `head()` et `tail()` : affichent les premières et dernières lignes.

```python
df.head()
```

```python
df.head(2)
```

```python
df.tail()
```

```python
df.tail(2)
```

- `info()` : affiche les informations générales du DataFrame.

```python
df.info()
```

- `describe()` : affiche un résumé statistique des données.

```python
df.describe(include="object")
```

## 4. Sélection de colonnes

- `df["message"]` : sélectionne une seule colonne (Series).
- `df[["message", "classe"]]` : sélectionne plusieurs colonnes (DataFrame).

```python
df["message"]
```

```python
df[["message","classe"]]
```

## 5. Filtrage des données

Le filtrage permet de sélectionner les lignes qui respectent une condition.

- `==` : compare deux valeurs.
- `&` : ET.
- `|` : OU.
- `value_counts()` : compte les occurrences de chaque valeur.

```python
df[df["classe"]=="spam"]
```

```python
df[df["classe"]=="normal"]
```

```python
df["message"].value_counts()
```

```python
df["classe"].value_counts()
```

## 6. Valeurs manquantes

Les valeurs manquantes sont des données absentes dans le DataFrame.

- `isna()` : détecte les valeurs manquantes.
- `isna().sum()` : les compte par colonne.
- `dropna()` : supprime les lignes concernées.
- `fillna()` : remplace les valeurs manquantes.

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df1=df.dropna()
df1[["message","classe"]]
```

```python
df["message"]=df["message"].fillna("pas de valeur")
df[["message","classe"]]
```

## 7. Création d'une nouvelle colonne

On peut ajouter une nouvelle colonne à un DataFrame en lui affectant une valeur.

La méthode `str.len()` permet de calculer le nombre de caractères de chaque message.

```python
df["longeur_message"]=df["message"].str.len()
df
```

```python
df.to_csv("../messages.csv", index=False)
```
