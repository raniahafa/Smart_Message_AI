**1. Variables contenant des messages**

```python
message_1 = "Salut, on se voit demain à 10h devant la fac ?"
message_2 = "GAGNEZ un iPhone 15 GRATUIT !!! Cliquez ici"
message_3 = "Ok"
```

**2. Fonction qui retourne la longueur d’un message**

```python
def longeur_message(texte):
  return len(texte)
print(longeur_message(message_1))
print(longeur_message(message_2))
print(longeur_message(message_3))
```

**3. Fonction qui retourne le nombre de mots dans une phrase**

```python
def compter_mot(texte):
  return len(texte.split())
print(message_1, " => contient", compter_mot(message_1), "mots")
print(message_2, " => contient", compter_mot(message_2), "mots")
print(message_3, " => contient", compter_mot(message_3), "mots")
```

**4. Liste de messages et boucle**

```python
messages = [
    "Salut, on se voit demain à 10h devant la fac ?",
    "GAGNEZ un iPhone 15 GRATUIT !!! Cliquez ici : http://exemple.com/prix",
    "N'oublie pas d'apporter le cahier de maths",
    "URGENT : votre compte sera bloqué, confirmez vos informations maintenant",
    "Merci pour ton aide hier, c'était super",
    "Félicitations ! Vous avez gagné 1000 euros. Répondez GAGNE pour recevoir votre prix",
    "Ok",
    "Tu peux m'envoyer le rapport avant ce soir ?",
    "Offre limitée : -70% sur tout le site, profitez-en vite !!!",
    "La réunion est reportée à jeudi 14h, salle B2",
    "Hello, how are you? Call me when you are free",
    "WINNER!! You have been selected for a FREE gift, claim now",
    "Je serai en retard de 10 minutes, désolée",
    "Cliquez sur ce lien pour obtenir votre cadeau gratuit : http://exemple.com/cadeau",
    "Bon anniversaire aya ! Passe une excellente journée",
]
print("le nombre de message de cette liste est:",len(messages))
for msg in messages :
  print(msg)
```

**Erreur 1 : oubli des guillemets**

Cette erreur se produit lorsqu'on écrit un texte sans guillemets. Python considère alors le texte comme un nom de variable et peut afficher une erreur `NameError`.

```python
message = Bonjour
print(message)
```

**Correction :** nous ajoutons des guillemets autour du texte pour que Python le reconnaisse comme une chaîne de caractères.

```python
message = "Bonjour"
print(message)
```

**Erreur 2 : indentation incorrecte**

Cette erreur se produit lorsque l'instruction d'une boucle n'est pas correctement décalée. Python affiche alors généralement une erreur `IndentationError`.

```python
for msg in messages:
print(msg)
```

**Correction :** nous ajoutons quatre espaces avant `print(message)` pour indiquer que cette instruction appartient à la boucle `for`.

```python
for msg in messages:
   print(msg)
```

**Erreur 3 : incompatibilité des types de données**

Cette erreur se produit lorsqu'on essaie d'additionner un nombre entier et une chaîne de caractères. Python affiche une erreur `TypeError` car ces deux types de données ne peuvent pas être additionnés directement.

```python
age = 20
message = "Mon âge est : "
print(message + age)
```

**Correction :** nous utilisons `str()` pour convertir le nombre en texte avant de le concaténer avec le message.

```python
age = 20
message = "Mon âge est : "
print(message + str(age))
```
