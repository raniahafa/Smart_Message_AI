# Semaine 3 : Acquisition et compréhension du dataset

**Objectif :** découvrir le dataset principal, comprendre son organisation et vérifier la qualité des données avant le Machine Learning.

## 1. Présentation du dataset

Le projet utilise le dataset **SMS Spam Collection**.

Il contient des messages SMS en anglais classés en deux catégories :
- `ham` : message normal.
- `spam` : message indésirable.

## 2.Chargement du dataset avec pandas 
- `sep="\t"` : définit la tabulation comme séparateur.
- `header=None` : indique l'absence d'en-tête.
- `names` : définit les noms des colonnes.

```python
import pandas as pd 
import csv
df=pd.read_csv("../data/raw/SMSSpamCollection",sep="\t" ,header=None,names=["label","message"],quoting=csv.QUOTE_NONE)
```

## 3.4 Répartition des classes

Nous comptons le nombre de messages Ham et Spam pour connaître la répartition des classes dans le dataset.

```python
df["label"].value_counts()
```

Avec normalize=True : on obtient la proportion de chaque classe par rapport au nombre total de messages.

```python
df["label"].value_counts(normalize=True)*100
```

**Observation :** Les messages Ham représentent environ 86,59 % du dataset, contre 13,41 % pour les messages Spam. Le dataset présente donc un déséquilibre des classes.

## 4. Vérification de la qualité des données
